```python
dangerously_pull_public_prompt=True
```

```python
from langsmith import Client

# Initialize the client
client = Client()

# Pull the prompt (e.g., a ReAct agent prompt)
prompt = client.pull_prompt("hwchase17/react-chat", dangerously_pull_public_prompt=True)

print(prompt)
```

```python
prompt.template
```

```python
# Step 3: Create the ReAct agent manually with the pulled prompt
agent = create_react_agent(
    llm=llm,
    tools=[search_tool],
    prompt=prompt
)
```

```python
# Step 4: Wrap it with AgentExecutor
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose=True
)
```

```python
# Step 5: Invoke
response = agent_executor.invoke({"input": "3 ways to reach goa from delhi", "chat_history": []})
print(response)
```

```python
response['output']
```

```python
response_search = agent_executor.invoke({"input": "What is the current weather in varanasi City?", "chat_history": []})
print(response_search)
```

```python
response_search['output']
```

```python
weatherstack_api_key = ""
```

```python
@tool
def get_weather_data(city: str) -> str:
  """
  This function fetches the current weather data for a given city
  """
  url = f'https://api.weatherstack.com/current?access_key={weatherstack_api_key}&query={city}'

  response = requests.get(url)

  return response.json()
```

```python
get_weather_data.invoke('Varanasi')
```
